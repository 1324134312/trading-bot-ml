# Data Exploration Notebook

This notebook helps you explore and analyze the collected market data.


```python
import sys
import os
sys.path.insert(0, os.path.dirname(os.path.dirname(os.path.abspath(''))))

import pandas as pd
import matplotlib.pyplot as plt
from src.data_collection.data_fetcher import DataFetcher
from config.settings import DEFAULT_TICKERS, RAW_DATA_DIR

# Initialize fetcher
fetcher = DataFetcher(data_dir=RAW_DATA_DIR)
```

```python
# Load and explore a specific ticker
ticker = 'AAPL'  # Change this to explore different stocks
data = fetcher.load_ticker_data(ticker)

if data is not None:
    print(f"Data for {ticker}:")
    print(data.head())
    print(f"\nShape: {data.shape}")
    print(f"Date range: {data['Date'].min()} to {data['Date'].max()}")
```

```python
# Plot price chart
if data is not None:
    fig, axes = plt.subplots(2, 1, figsize=(12, 8))
    
    # Price chart
    axes[0].plot(data['Date'], data['Close'], label='Close Price')
    axes[0].set_title(f'{ticker} Close Price')
    axes[0].set_xlabel('Date')
    axes[0].set_ylabel('Price ($)')
    axes[0].legend()
    axes[0].grid(True)
    
    # Volume chart
    axes[1].bar(data['Date'], data['Volume'], alpha=0.5)
    axes[1].set_title(f'{ticker} Trading Volume')
    axes[1].set_xlabel('Date')
    axes[1].set_ylabel('Volume')
    axes[1].grid(True)
    
    plt.tight_layout()
    plt.show()
```

```python
# Calculate basic statistics
if data is not None:
    print(f"\n{ticker} Statistics:")
    print(f"Average Close Price: ${data['Close'].mean():.2f}")
    print(f"Price Range: ${data['Close'].min():.2f} - ${data['Close'].max():.2f}")
    print(f"Average Volume: {data['Volume'].mean():,.0f}")
    
    # Calculate returns
    data['Returns'] = data['Close'].pct_change()
    print(f"\nAverage Daily Return: {data['Returns'].mean()*100:.4f}%")
    print(f"Daily Volatility (std): {data['Returns'].std()*100:.4f}%")
```
