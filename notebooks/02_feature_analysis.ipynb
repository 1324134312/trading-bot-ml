Explore and analyze the engineered features for your trading bot.


```python
import sys
import os
sys.path.insert(0, os.path.dirname(os.path.dirname(os.path.abspath(''))))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load processed data
ticker = 'AAPL'
data = pd.read_csv(f'../data/processed/{ticker}_features.csv', index_col='Date', parse_dates=True)

print(f"📊 Data shape: {data.shape}")
print(f"📈 Columns: {len(data.columns)}")
print(f"📅 Date range: {data.index.min()} to {data.index.max()}")
```

## Visualize Technical Indicators


```python
# Plot price with moving averages and Bollinger Bands
fig, axes = plt.subplots(4, 1, figsize=(15, 12))

# Price with MAs
axes[0].plot(data.index, data['Close'], label='Close Price', linewidth=2)
axes[0].plot(data.index, data['SMA_20'], label='SMA 20', alpha=0.7)
axes[0].plot(data.index, data['SMA_50'], label='SMA 50', alpha=0.7)
axes[0].plot(data.index, data['BB_Upper'], label='BB Upper', color='red', alpha=0.3)
axes[0].plot(data.index, data['BB_Lower'], label='BB Lower', color='green', alpha=0.3)
axes[0].fill_between(data.index, data['BB_Lower'], data['BB_Upper'], alpha=0.1)
axes[0].set_title(f'{ticker} Price with Technical Indicators')
axes[0].set_ylabel('Price ($)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# RSI
axes[1].plot(data.index, data['RSI_14'], label='RSI', color='purple')
axes[1].axhline(y=70, color='r', linestyle='--', alpha=0.5, label='Overbought')
axes[1].axhline(y=30, color='g', linestyle='--', alpha=0.5, label='Oversold')
axes[1].set_title('RSI (Relative Strength Index)')
axes[1].set_ylabel('RSI')
axes[1].set_ylim(0, 100)
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# MACD
axes[2].plot(data.index, data['MACD'], label='MACD', color='blue')
axes[2].plot(data.index, data['MACD_Signal'], label='Signal', color='orange')
axes[2].bar(data.index, data['MACD_Histogram'], label='Histogram', alpha=0.3)
axes[2].set_title('MACD')
axes[2].set_ylabel('MACD')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

# Volume
axes[3].bar(data.index, data['Volume'], alpha=0.5, label='Volume')
axes[3].plot(data.index, data['Volume_MA'], label='Volume MA', color='red')
axes[3].set_title('Trading Volume')
axes[3].set_ylabel('Volume')
axes[3].legend()
axes[3].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Analyze Feature Correlations with Target


```python
# Select key features to analyze
key_features = [
    'RSI_14', 'RSI_7',
    'MACD', 'MACD_Histogram', 'MACD_Signal',
    'Stoch_K', 'Stoch_D',
    'BB_Position', 'BB_Width',
    'ADX_14', 'ATR_14',
    'Volume_Ratio',
    'Price_Change', 'Price_Change_5', 'Price_Change_10'
]

# Get correlation with target
if 'Target_Direction' in data.columns:
    corr_with_target = data[key_features + ['Target_Direction']].corr()['Target_Direction'].sort_values(ascending=False)
    corr_with_target = corr_with_target.drop('Target_Direction')
    
    print("🔍 Correlation with Target (Next Day Direction):")
    print(corr_with_target)
    
    # Visualize
    plt.figure(figsize=(10, 6))
    corr_with_target.plot(kind='barh')
    plt.title('Feature Correlation with Target Direction')
    plt.xlabel('Correlation')
    plt.tight_layout()
    plt.show()
```
