# Model Predictions Notebook

Use the trained ML model to make predictions on stock price direction.


```python
import sys
import os
sys.path.insert(0, os.path.dirname(os.path.dirname(os.path.abspath(''))))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.models.predictor import ModelPredictor

# Load trained model
model_path = '../src/models/trained_models/multi_ticker_xgb_model.pkl'
predictor = ModelPredictor(model_path)

print("✅ Model loaded successfully!")
print(f"📋 Model uses {len(predictor.trainer.feature_names)} features")
```

## Make Predictions on Test Data


```python
# Load test data
ticker = 'AAPL'
data = pd.read_csv(f'../data/processed/{ticker}_features.csv')

# Make predictions
predictions = predictor.predict_direction(data.copy())

print(f"📊 Predictions for {ticker}:")
print(f"   Total predictions: {len(predictions)}")
print(f"   Up predictions:    {sum(predictions['Prediction'] == 1)}")
print(f"   Down predictions:  {sum(predictions['Prediction'] == 0)}")
print(f"   Average confidence: {predictions['Confidence'].mean():.2%}")

# Show recent predictions
recent = predictions[['Date', 'Close', 'Prediction', 'Prediction_Prob', 'Confidence', 'Target_Direction']].tail(10)
print("\n📈 Recent Predictions:")
print(recent.to_string(index=False))
```

## Find Buy/Sell Signals


```python
# Get buy and sell signals with high confidence
buy_signals = predictor.get_buy_signals(predictions, min_confidence=0.7)
sell_signals = predictor.get_sell_signals(predictions, min_confidence=0.7)

print(f"🟢 Buy Signals (High Confidence): {len(buy_signals)}")
print(f"🔴 Sell Signals (High Confidence): {len(sell_signals)}")

if len(buy_signals) > 0:
    print("\n🟢 Recent Buy Signals:")
    print(buy_signals[['Date', 'Close', 'Prediction_Prob', 'Confidence']].tail(5))
    
if len(sell_signals) > 0:
    print("\n🔴 Recent Sell Signals:")
    print(sell_signals[['Date', 'Close', 'Prediction_Prob', 'Confidence']].tail(5))
```
